```python
import pandas as pd

gold = pd.read_pickle("data/golden69_dataset.pkl")

print(gold.columns)
print(gold.head(3))
```

```python
corpus = pd.read_pickle("big_data/10k_sample_over_all.pkl")

print(corpus.columns)
def normalize_pmcid(x):
    if pd.isna(x):
        return None
    x = str(x)
    return x.replace("PMC", "").strip()

```

```python
gold_ids = set(gold["pmcid"].apply(normalize_pmcid))
corpus_ids = set(corpus["pmcid"].apply(normalize_pmcid))
```

```python
found = gold_ids.intersection(corpus_ids)
missing = gold_ids.difference(corpus_ids)

print(f"Total GOLD: {len(gold_ids)}")
print(f"Found in corpus: {len(found)}")
print(f"Missing: {len(missing)}")

coverage = len(found) / len(gold_ids)
print(f"Coverage: {coverage:.2%}")
```
