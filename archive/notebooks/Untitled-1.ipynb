```python
import requests
import time
import os

# Deine Liste der pubids aus dem PubMedQA Datensatz (Beispiel-IDs)
 
pubids = dfqa['pubid']
# Ordner für die XML-Downloads erstellen
output_dir = "pubmed_xmls"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

def download_pubmed_xmls(id_list):
    # 1. Schritt: PMIDs in PMCIDs umwandeln
    # Wir fragen in Batches ab (max 200 IDs pro Request empfohlen)
    id_string = ",".join(id_list)
    conv_url = f"https://www.ncbi.nlm.nih.gov{id_string}&format=json&tool=my_tool&email=my_email@example.com"
    
    print("Prüfe Verfügbarkeit von Open Access Volltexten...")
    response = requests.get(conv_url).json()
    
    pmc_ids = []
    for record in response.get('records', []):
        if 'pmcid' in record:
            pmc_ids.append((record['pmid'], record['pmcid']))
    
    print(f"Gefunden: {len(pmc_ids)} von {len(id_list)} Artikeln haben einen freien Volltext.\n")

    # 2. Schritt: XML für jede gefundene PMCID herunterladen
    for pmid, pmcid in pmc_ids:
        print(f"Lade XML für {pmcid} (PMID: {pmid}) herunter...")
        
        # E-Utilities Fetch URL
        fetch_url = f"https://eutils.ncbi.nlm.nih.gov{pmcid}&retmode=xml"
        
        try:
            xml_data = requests.get(fetch_url)
            if xml_data.status_code == 200:
                filename = os.path.join(output_dir, f"{pmcid}_{pmid}.xml")
                with open(filename, 'wb') as f:
                    f.write(xml_data.content)
                print(f"Gespeichert: {filename}")
            else:
                print(f"Fehler beim Download von {pmcid}")
        except Exception as e:
            print(f"Fehler bei {pmcid}: {e}")
        
        # Wichtig: Kurze Pause, um die NCBI-Server nicht zu überlasten (Rate Limiting)
        time.sleep(0.4)

if __name__ == "__main__":
    download_pubmed_xmls(pubids)
```

```python
import pandas as pd
import requests
import io

# 1. Lade die offizielle PMC Open Access Dateiliste (ca. 30MB)
# Diese Liste enthält: File-Pfad, Citation, Accession ID (PMCID), PMID, etc.
print("Lade Open Access Referenzliste von PMC...")
oa_list = pd.read_csv('../../PubMed/Supplements/oa_file_list.csv') 

# 2. Deine pubids aus dem PubMedQA Datensatz (als Beispiel)
# Ersetze dies durch: my_pubids = df_pubmedqa['pubid'].tolist()
#my_pubids = [21645374, 16418930, 9488747] 
dataset = load_dataset("pubmed_qa", "pqa_labeled", split="train")
df_qa = pd.DataFrame(dataset)
my_pubids = set(df_qa['pubid'].astype(int))  # PMIDs als Strings
print(f"Anzahl PubMedQA-Papers: {len(pmids_qa)}")


# 3. Filtere die OA-Liste nach deinen pubids
# Wichtig: PMID in der Liste ist oft Float, daher konvertieren
oa_list['PMID'] = oa_list['PMID'].fillna(0).astype(int)
matches = oa_list[oa_list['PMID'].isin(my_pubids)]

print(f"Treffer: {len(matches)} von {len(my_pubids)} sind Open Access.")

# 4. Download der XMLs für die Treffer
# Wir nutzen hier den stabilen Web-Link, der direkt aus der Accession ID generiert wird
for _, row in matches.iterrows():
    pmcid = row['Accession ID']
    pmid = row['PMID']
    
    # Der direkte Pfad zum XML via E-Link (sehr stabil für OA-Content)
    xml_url = f"https://www.ncbi.nlm.nih.gov{pmcid}/node/body/xml/"
    # Alternativ über eutils:
    # xml_url = f"https://eutils.ncbi.nlm.nih.gov{pmcid}&retmode=xml"

    print(f"Lade {pmcid} (PMID {pmid})...")
    r = requests.get(xml_url)
    
    if r.status_code == 200 and len(r.content) > 500: # Validierung: Nicht leer
        with open(f"{pmcid}.xml", "wb") as f:
            f.write(r.content)
    else:
        print(f"Download fehlgeschlagen oder kein Volltext-XML für {pmcid}")
```

```python
from datasets import load_dataset
import pandas as pd
import requests
import time
from pathlib import Path
import urllib.request

# ------------------------------------------------------------
# 1. PubMedQA-Fragen laden (nur die manuell annotierten)
# ------------------------------------------------------------
dataset = load_dataset("pubmed_qa", "pqa_labeled", split="train")
df_qa = pd.DataFrame(dataset)
pmids_qa = set(df_qa['pubid'].astype(str))  # PMIDs als Strings
print(f"Anzahl PubMedQA-Papers: {len(pmids_qa)}")

# ------------------------------------------------------------
# 2. PMC Open Access Subset laden
# ------------------------------------------------------------
print("Lade PMC OA non-commercial Liste...")
df_oa = pd.read_csv('../../PubMed/Supplements/oa_file_list.csv') 

# PMID-Spalte bereinigen (führende/nachgestellte Punkte entfernen, Komma-getrennte Werte vorbereiten)
df_oa['PMID'] = df_oa['PMID'].astype(str).str.replace(r'\.0$', '', regex=True)

# ------------------------------------------------------------
# 3. Mapping von PMID -> Accession ID erstellen
#    (da eine Zeile mehrere PMIDs enthalten kann)
# ------------------------------------------------------------
pmid_to_accession = {}
for idx, row in df_oa.iterrows():
    pmids = str(row['PMID']).split(',')
    accession = row['Accession ID']
    for p in pmids:
        p_clean = p.strip()
        if p_clean:  # leere Einträge ignorieren
            pmid_to_accession[p_clean] = accession

print(f"Anzahl geladener PMIDs aus OA-Liste: {len(pmid_to_accession)}")

# ------------------------------------------------------------
# 4. Schnittmenge mit PubMedQA-Papers bestimmen
# ------------------------------------------------------------
gemeinsame_pmids = pmids_qa & set(pmid_to_accession.keys())
print(f"Gemeinsame Papers: {len(gemeinsame_pmids)}")

# ------------------------------------------------------------
# 5. Kombiniertes DataFrame erstellen (QA-Daten + Accession ID)
# ------------------------------------------------------------
combined_records = []

for pmid in gemeinsame_pmids:
    # Zeile aus df_qa holen (sollte genau eine sein)
    qa_row = df_qa[df_qa['pubid'].astype(str) == pmid]
    if qa_row.empty:
        print(f"Warnung: PMID {pmid} nicht in df_qa gefunden – überspringe.")
        continue
    qa_dict = qa_row.iloc[0].to_dict()
    
    # Accession ID aus Mapping holen
    accession = pmid_to_accession.get(pmid)
    if accession is None:
        print(f"Warnung: Keine Accession ID für PMID {pmid} gefunden – setze auf None.")
    
    qa_dict['Accession ID'] = accession
    combined_records.append(qa_dict)

# DataFrame aus den gesammelten Einträgen erstellen
combined_df = pd.DataFrame(combined_records)
print(f"Kombiniertes DataFrame enthält {len(combined_df)} Einträge.")
print(combined_df.head())

# Optional: DataFrame speichern
combined_df.to_csv("../../PubMed/QA_Papers_with_Accession.csv", index=False)
combined_df.to_pickle("../../PubMed/QA_Papers_with_Accession.pkl")
# ------------------------------------------------------------
# 6. XML-Download der gemeinsamen Papers (wie gehabt)
# ------------------------------------------------------------
download_dir = Path("../../PubMed/QA_Paper_OA")
download_dir.mkdir(exist_ok=True)

for pmid in gemeinsame_pmids:
    # Finde die Zeile in df_oa, die diese PMID enthält (für den FTP-Pfad)
    # Achtung: str.contains ist unscharf – besser mit dem Mapping arbeiten, aber hier bleibt es erstmal so
    row = df_oa[df_oa['PMID'].str.contains(pmid, na=False)]
    if row.empty:
        print(f"Keine OA-Information für PMID {pmid} gefunden (überspringe Download).")
        continue
    ftp_path = row.iloc[0]['File']
    file_url = f"ftp://ftp.ncbi.nlm.nih.gov/pub/pmc/{ftp_path}"
    local_file = download_dir / ftp_path.split('/')[-1]
    
    if not local_file.exists():
        print(f"Download {file_url} ...")
        try:
            urllib.request.urlretrieve(file_url, local_file)
            print(f"  -> gespeichert als {local_file}")
        except Exception as e:
            print(f"  FEHLER bei {file_url}: {e}")
    else:
        print(f"  {local_file} existiert bereits, überspringe.")
    
    time.sleep(0.5)
```
